```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
x = housing["data"]
y = housing["target"]
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size= 0.2 , random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scaling = StandardScaler()
x_train_transformed = scaling.fit_transform(x_train)
x_test_transformed = scaling.transform(x_test)
```

```python
from sklearn.svm import LinearSVR

lin_svr = LinearSVR(random_state= 42)
lin_svr.fit(x_train_transformed , y_train)
```

```python
from sklearn.metrics import mean_squared_error
y_pred = lin_svr.predict(x_train_transformed)
mse = mean_squared_error(y_train , y_pred)
mse
```

```python
import numpy as np

np.sqrt(mse)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVR
from scipy.stats import reciprocal , uniform

param_distribution = {"gamma" : reciprocal(0.001 , 0.1) , "C" : uniform(1 , 10)}
best_fit = RandomizedSearchCV(SVR() , param_distribution ,cv= 2 , n_jobs= -1 )
best_fit.fit(x_train_transformed , y_train)
```

```python
best_fit.best_estimator_
```

```python
y_pred_1 = best_fit.best_estimator_.predict(x_train_transformed)
mse = mean_squared_error(y_train , y_pred_1)
np.sqrt(mse)
```

```python
#Test Value
y_pred_test = best_fit.best_estimator_.predict(x_test_transformed)
mse_test = mean_squared_error(y_test , y_pred_test)
np.sqrt(mse_test)
```
