```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron

iris = load_iris()
x = iris.data[: , 2:4] 
y = (iris.target == 0)
```

```python
per_clf = Perceptron(random_state= 42)
per_clf.fit(x , y)
```

```python
x_new = [[2 , 0.5] , [3 , 1.0]]
y_new = per_clf.predict(x_new)
print(y_new)
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

housing = fetch_california_housing()
x_train , x_test , y_train , y_test = train_test_split(housing.data , housing.target, random_state=42)
```

```python
ml_reg = MLPRegressor(hidden_layer_sizes=[50, 50 , 50] , verbose=True , early_stopping=True , random_state=42)
pipeline = make_pipeline(StandardScaler() , ml_reg)
pipeline.fit(x_train  , y_train)
```

```python
from sklearn.metrics import root_mean_squared_error
y_pred =pipeline.predict(x_test)
mse = root_mean_squared_error(y_test , y_pred)
mse
```

```python
#MLPClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import fetch_openml

mnist_fashion = fetch_openml(name='Fashion-MNIST' , as_frame=False)
target = mnist_fashion.target.astype(int)
```

```python
#train_test split
x_train , y_train = mnist_fashion.data[:60000] , target[:60000]
x_test , y_test = mnist_fashion.data[60000:] , target[60000:]
```

```python
import matplotlib.pyplot as plt

x_someimage = x_train[0].reshape(28 , 28)
plt.imshow(x_someimage , cmap="binary")
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat", 
                "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
plt.axis('off')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import make_pipeline

mlp_class = MLPClassifier(
    hidden_layer_sizes=[300 , 100],
    verbose=True,
    early_stopping=True,
    random_state=42
)

pipeline = make_pipeline(MinMaxScaler(),
                         mlp_class)

pipeline.fit(x_train , y_train)
```

```python
accuracy = pipeline.score(x_test , y_test)
accuracy
```
